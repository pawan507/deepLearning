```python
#importing necessary libraries
import torch 
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
```

```python
#constructing a neual network model
class SimpleNN(nn.Module):
    def __init__(self,input_size, hidden_size, output_size):
        super(SimpleNN, self).__init__()
        self.layer1 = nn.Linear(input_size, hidden_size)  # Input layer to hidden layer
        self.layer2 = nn.Linear(hidden_size, output_size)  # Hidden layer to output layer

    def forward(self, x):
        output=self.layer1(x)  # Pass through the first layer
        output=torch.relu(output)  # Apply ReLU activation function
        output=self.layer2(output)  # Pass through the second layer 
        return output
```
